```python
# ============================================================
# IMAGE GENERATION USING GENERATIVE ADVERSARIAL NETWORK (GAN)
# StyleGAN2-ADA - Complete Google Colab Code
# ============================================================

# 1. Clear previous failed compilation cache, set explicit build directory, and install libraries
import os
import shutil

# Set an alternate writeable directory for torch compilation to bypass path/permission issues
os.environ["TORCH_EXTENSIONS_DIR"] = "/tmp/torch_extensions"

if os.path.exists("/tmp/torch_extensions"):
    print("Clearing previous compilation cache...")
    shutil.rmtree("/tmp/torch_extensions")

!pip install -q torchmetrics pytorch-fid

# 2. Import libraries
import sys
import subprocess
import urllib.request
import torch
import matplotlib.pyplot as plt

from torchvision import datasets as dsets
from torchvision import transforms
from torchvision.utils import save_image
from torch.utils.data import DataLoader, TensorDataset
from torchmetrics.image.inception import InceptionScore

# 3. Select device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# ============================================================
# 4. Clone StyleGAN2-ADA repository
# ============================================================

REPO = "stylegan2-ada-pytorch"
REPO_URL = "https://github.com/NVlabs/stylegan2-ada-pytorch.git"

if not os.path.exists(REPO):
    print("Cloning StyleGAN2-ADA...")
    subprocess.run(["git", "clone", REPO_URL], check=True)

sys.path.append(os.path.abspath(REPO))

print("StyleGAN2-ADA repository ready!")

# ============================================================
# 5. Download pretrained FFHQ model
# ============================================================

MODEL_URL = "https://nvlabs-fi-cdn.nvidia.com/stylegan2-ada-pytorch/pretrained/ffhq.pkl"
MODEL_PATH = "ffhq.pkl"

if not os.path.exists(MODEL_PATH):
    print("Downloading pretrained FFHQ model...")
    urllib.request.urlretrieve(MODEL_URL, MODEL_PATH)

print("FFHQ model ready!")

# ============================================================
# 6. Load StyleGAN2-ADA generator
# ============================================================

import legacy
import dnnlib

with open(MODEL_PATH, "rb") as f:
    G = legacy.load_network_pkl(f)["G_ema"].to(device)

print("Generator loaded successfully!")
print("Image resolution:", G.img_resolution)

# ============================================================
# 7. Generate 10 synthetic images
# ============================================================

NUM_IMAGES = 10

os.makedirs("generated_images", exist_ok=True)

imgs = []

print("\nGenerating images...")

for i in range(NUM_IMAGES):
    z = torch.randn([1, G.z_dim], device=device)
    label = torch.zeros([1, G.c_dim], device=device)

    with torch.no_grad():
        img = G(
            z,
            label,
            truncation_psi=0.5,
            noise_mode="const"
        )

    img = ((img + 1) * 0.5).clamp(0, 1)

    save_image(
        img,
        f"generated_images/fake_{i}.png"
    )

    imgs.append(img.cpu())

imgs = torch.cat(imgs)

print("10 synthetic images generated!")

# ============================================================
# 8. Display generated images
# ============================================================

fig, axes = plt.subplots(2, 5, figsize=(15, 6))

for i, ax in enumerate(axes.flat):
    img = imgs[i].permute(1, 2, 0).numpy()
    ax.imshow(img)
    ax.set_title(f"Generated {i+1}")
    ax.axis("off")

plt.tight_layout()
plt.show()

# ============================================================
# 9. Calculate Inception Score
# ============================================================

print("\nCalculating Inception Score...")

dataset_is = TensorDataset(
    (imgs * 255).clamp(0, 255).to(torch.uint8)
)

loader_is = DataLoader(
    dataset_is,
    batch_size=2
)

is_metric = InceptionScore(normalize=False)

for batch in loader_is:
    is_metric.update(batch[0])

mean, std = is_metric.compute()

print(f"Inception Score: {mean:.2f} ± {std:.2f}")

# ============================================================
# 10. Download and prepare CelebA real images
# ============================================================

print("\nPreparing real CelebA images...")

transform = transforms.Compose([
    transforms.Resize((512, 512)),
    transforms.ToTensor()
])

try:
    celeb_dataset = dsets.CelebA(
        root="./data",
        split="train",
        transform=transform,
        download=True
    )

    real_loader = DataLoader(
        celeb_dataset,
        batch_size=1,
        shuffle=True
    )

    os.makedirs("real_images", exist_ok=True)

    for i, (img, _) in enumerate(real_loader):
        if i >= NUM_IMAGES:
            break

        save_image(
            img,
            f"real_images/real_{i}.png"
        )

    print("10 real CelebA images prepared!")

    # ========================================================
    # 11. Calculate FID
    # ========================================================

    print("\nCalculating FID...")

    result = subprocess.run(
        [
            sys.executable,
            "-m",
            "pytorch_fid",
            "real_images",
            "generated_images"
        ],
        capture_output=True,
        text=True
    )

    print(result.stdout)

    if result.stderr:
        print(result.stderr)

except Exception as e:
    print("\nCelebA/FID could not be completed.")
    print("Error:", e)

# ============================================================
# FINAL RESULT
# ============================================================

print("\n" + "=" * 50)
print("       GAN EXPERIMENT RESULT")
print("=" * 50)

print("Model            : StyleGAN2-ADA")
print("Dataset          : FFHQ / CelebA")
print("Generated Images : 10")
print(f"Inception Score  : {mean:.2f} ± {std:.2f}")

print("FID              : Check output above")

print("=" * 50)
print("Experiment completed!")
```
